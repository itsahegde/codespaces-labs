In probability theory, the birthday problem asks for the probability that, in a set of n randomly chosen people, at least two will share the same birthday.

The probability that atleast 2 people will have the same birthday let be PA. We will first calculate the probability of this first from 2 to some number n with a graph subsequently.

Before this we shall calculate the probability of no 2 people sharing the same birthday.

```python
def prob(n):
  temp = 365
  res = 1
  for i in range(1, n+1):
    res*=(temp/365.0)
    temp-= 1
  return res
```

```python
ans = prob(2)
print(ans)

#probability of 2 people sharing birthday is 1 - ans

sol = 1 - ans
print(sol)
```

```python
def record(n):
    res = []
    for i in range(1, n+1):
        res.append(prob(i))
    return res
def record_inverse(n):
    res = []
    for i in range(1, n+1):
        res.append(1 - prob(i))
    return res
```

```python
res = record(100)
print(res)
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def plot_graph(num):
  res = record(num)
  sol = record_inverse(num)

  y_1 = np.array(res)
  x_1 = np.array(range(1, num+1))

  y_2 = np.array(sol)
  x_2 = np.array(range(1, num+1))

  plt.xlabel("Number of people")
  plt.ylabel("Probability")

  plt.plot(x_1, y_1, x_2, y_2)
  plt.xticks(np.arange(1, num+1, 1.0))
  plt.yticks(np.arange(0, 1.1, 0.1))
  plt.show()
```

```python
plot_graph(25)
```
